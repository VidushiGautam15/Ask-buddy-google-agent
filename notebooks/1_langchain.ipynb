###STATIC PROMPTS


```python
from langchain_groq import ChatGroq 
llm = ChatGroq(model="openai/gpt-oss-120b")
```

```python
prompts=[
    ("system", "You are a helpful assistant."),
    ("user", "Write a poem about the beauty of nature.")
]
res=llm.invoke(prompts)
print(res.content)
```
