```python
from dotenv import load_dotenv
load_dotenv()  # Load environment variables from .env file
```

```python
import langchain_core.tools as tools
```

```python
@tools.tool 
def add_numbers(a: int, b: int) -> int:
    """Add two numbers."""
    return a + b
```

```python
add_numbers.invoke({"a": 3, "b": 5})  # This will return 8
```

```python
from langchain_groq import ChatGroq
from langchain.agents import create_agent
```

```python
llm = ChatGroq(model="openai/gpt-oss-120b")
```

```python
agent = create_agent(
    model=llm,
    tools=[add_numbers],
)
system_prompt = "You are a helpful assistant that can perform basic arithmetic operations."
```

```python
response=agent.invoke(
    {"messages": [{"role": "system", "content": system_prompt}, {"role": "user", "content": "What is the sum of 10 and 15?"}]}
)
```

```python
print(response["messages"][-1].content)  # This will print the response from the agent, which should include the result of the addition.
```
