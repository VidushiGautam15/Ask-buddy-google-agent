```python
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(model="gemini-3.8-flash")
```

```python
prompts = [
    {"role": "system", "content": "You are a JavaScript developer."},
    {"role": "user", "content": """
Explain the difference between map and filter.

Give the answer in a clear paragraph format.
Use separate paragraphs for each concept.
Do not write the entire answer in one line.
Use proper line breaks and spacing.
"""}
]

res=llm.invoke(prompts) 
print(res.content)
```

###PROMPTS TEMPLATE - DYNAMIC PROMPT

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate

llm = ChatGoogleGenerativeAI(model="gemini-3.8-flash")
```

```python
prompts=ChatPromptTemplate.from_messages([
    {"role": "system", "content": "You are a translater which translates the input in {language}."},
    {"role": "user", "content": "{query}"}
])
```

```python
final_prompt=prompts.format_messages(
    language="hinglish",
    query="I hate coding in JavaScript."
)

res=llm.invoke(final_prompt)
print(res.content)
```

###CHAINS - RUNNABLES

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate

llm = ChatGoogleGenerativeAI(model="gemini-3.8-flash")
```

```python
prompts=ChatPromptTemplate.from_messages([
    {"role": "system", "content": "You are a translater which translates the input in {language}."},
    {"role": "user", "content": "{query}"}
])
```

```python
chains = prompts | llm

res = chains.invoke({
    "language": "french",
    "query": "I hate coding in JavaScript."
})

print(res.content)
```

###LLM OUTPUT

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

llm = ChatGoogleGenerativeAI(model="gemini-3.8-flash")
out=StrOutputParser()
```

```python
prompts=ChatPromptTemplate.from_messages([
    {"role": "system", "content": "You are a translater which translates the input in {language}."},
    {"role": "user", "content": "{query}"}
])
```

```python
chains = prompts | llm | out

res = chains.invoke(){
    "language": "french",
    "query": "I hate coding in JavaScript."
}

print(res)
```
