### SOLUTION IS:STRUCTURE OUTPUT

```python
from pydantic import BaseModel, Field
from typing import List
```

```python
class response_model(BaseModel):
    name: str
    email: str
    age: int
    city: str
    
```

```python
structured_llm=llm.with_structured_output(response_model)
res=structured_llm.invoke(f"Extract the name, email address, age, and city from the following text: {text}")
```

```python
class movie_model(BaseModel):
    title: str=Field(..., description="The title of the movie")
    director: str=Field(..., description="The director of the movie")
    year: int=Field(..., description="The year the movie was released")
    genre: str=Field(..., description="The genre of the movie")
```

```python
class AllMovies(BaseModel):
    movies: list[movie_model]
```

```python
movie_llm=llm.with_structured_output(AllMovies)
```

```python
res= movie_llm.invoke("give me three recent movies with their title, director, year, and genre in a structured format")
```
